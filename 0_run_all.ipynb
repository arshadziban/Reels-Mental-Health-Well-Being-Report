# Run all steps
Runs each notebook in order. If a step fails, it stops there and the error is saved in that notebook.

```python
import os
import time
import nbformat
from nbclient import NotebookClient

BASE = os.getcwd()  # project root (notebook's folder)


def run_notebook(name):
    """Execute a notebook top to bottom and save its outputs back into the file."""
    path = os.path.join(BASE, name)
    print(f'Running {name} ...')
    start = time.time()
    nb = nbformat.read(path, as_version=4)
    client = NotebookClient(nb, kernel_name='python3', timeout=None,
                            resources={'metadata': {'path': BASE}})
    try:
        client.execute()
    finally:
        nbformat.write(nb, path)  # keep outputs, including the error if one happened
    print(f'Done {name} in {time.time() - start:.1f}s')
```

## Step 1: Fine-tune dataset

```python
run_notebook('1_finetune_dataset.ipynb')
```

## Step 2: Fix groups

```python
run_notebook('2_fix_group.ipynb')
```

## Step 3: Check outliers

```python
run_notebook('3_check_outlier.ipynb')
```
