```python
import os
import pandas as pd
from dotenv import load_dotenv
from gsheet_loader import load_sheet
load_dotenv()
```

```python
# Load a Google Sheet
url = os.environ["GOOGLE_SHEET_URL"]
data = load_sheet(url)
df = pd.read_csv(data)
df.head()
```

```python
# Rename columns to short, code-based names (e.g. A1_age, A2_gender, ...)
import re

# Human-readable label for each question code (based on the English question text)
code_labels = {
    "A1": "age",
    "A2": "gender",
    "A3": "religion",
    "A4": "marital_status",
    "A5": "residence",
    "A6": "occupation",
    "B1": "family_type",
    "B2": "family_time_hours",
    "B3": "family_relationship",
    "B4": "in_person_feeling",
    "C1": "height",
    "C2": "weight",
    "C3": "sitting_duration",
    "C4": "physical_problems",
    "D1": "years_watching",
    "D2": "daily_watch_time",
    "D3": "peak_watch_time",
    "D4": "multitasking_while_watching",
    "E1": "sleep_hours",
    "E2": "sleep_trouble",
    "E3": "exercise_involvement",
    "E4": "substance_use",
    "F1": "attention_decreased",
    "F2": "memory_reduced",
    "F3": "learning_interest_decreased",
    "F4": "performance_affected",
    "G1": "mood_swings",
    "G2": "fomo_anxiety",
    "G3": "social_comparison",
    "G4": "social_isolation",
    "G5": "mental_stress",
    "G6": "panic_attacks",
    "G7": "behavior_type",
    "G8": "addictive_watching",
    "H1": "mental_state_change",
    "H2": "content_type_effect",
    "H3": "distraction_after_viral",
    "H4": "brain_pressure",
    "H5": "sleep_quality_change",
    "H6": "emotional_reels_focus_loss",
    "H7": "curiosity_driven_rewatch",
}

code_pattern = re.compile(r"^\s*([A-H]\d{1,2})\s*[.)]")


def clean_column_name(col: str) -> str:
    if col.strip().lower() == "timestamp":
        return "timestamp"
    match = code_pattern.match(col)
    if match:
        code = match.group(1)
        label = code_labels.get(code)
        if label:
            return f"{code}_{label}"
    return col


df = df.rename(columns=clean_column_name)
df = df.drop(columns=["timestamp"])
df.columns.tolist()
```

```python
df.head()
```

```python
# Remove Bangla text, keeping only the English part of each value
# Values look like "English text (বাংলা text)" -> keep "English text"
bangla_pattern = re.compile(r"\s*\([^()]*[ঀ-৿][^()]*\)")


def strip_bangla(value):
    if isinstance(value, str):
        cleaned = bangla_pattern.sub("", value)
        cleaned = re.sub(r"[ঀ-৿]+", "", cleaned)
        return cleaned.strip()
    return value


df = df.apply(lambda col: col.map(strip_bangla))
df.head()
```

```python
# Save the cleaned dataset
os.makedirs("dataset", exist_ok=True)
df.to_csv("dataset/dataset_v1.csv", index=False)
```
